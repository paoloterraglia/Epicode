```python
filtro_anno = (df["date"].str.contains("2022"))& (df["location"]=="Italy")
query_anno=df.loc[filtro_anno]
display(query_anno)
```

```python
filtro_tp = ( (df["location"]=="Italy") | (df["location"]=="Germany") | (df["location"]=="France") ) & (df["date"] >= "2022-05-01") & (df["date"] <= "2023-04-30")
```

```python
query_tp=df.loc[filtro_tp]
display(query_tp)
```

```python
fig=px.box(data_frame=query_tp,x="location",y="icu_patients",title="Pazienti terapia intensiva - Italia,Germania,Francia(Maggio 2022 - Aprile 2023)")
```

```python
fig.show()
```

```python
## Commento 4.b 
##Dal boxplot si osserva che la Francia ha registrato il maggior numero di pazienti in terapia intensiva 
## nel periodo maggio 2022 - aprile 2023, seguita dalla Germania e infine dall'Italia che mostra 
## i valori più bassi. La Francia presenta anche una maggiore variabilità nei dati 
```

```python
filtro_osp = ( (df["location"]=="Italy") | (df["location"]=="Germany") | (df["location"]=="France")|(df["location"]=="Spain" )) & (df["date"] >= "2021-01-01") & (df["date"] <= "2021-12-31")
query_osp=df.loc[filtro_osp]
display(query_osp)
```

```python
somma_pazienti_osp = query_osp.groupby("location")["hosp_patients"].sum()
```

```python
display(somma_pazienti_osp)
```

```python
fig=px.bar(data_frame=somma_pazienti_osp.reset_index(),x="location",y="hosp_patients",title="Pazienti ospedalizzati - Italia, Germania, Francia e Spagna (Anno 2021)")
```

```python
fig.show()
```

```python
## 5.b Verifico i valori nulli per ogni nazione
display(query_osp[query_osp["hosp_patients"].isna()].groupby("location").size())
```
